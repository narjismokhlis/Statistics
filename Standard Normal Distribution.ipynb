The standard normal distribution is a normal distribution with mean 1 and variance 0. To generate a random sample from this distibution we use the R function rnorm(), but without specifying the mean and variance.
The code below generates and plots a radnom sample of size 100000.

```r
mysample<-rnorm(100000)
br<-seq(-4.7, 4.4, 0.1)
hist(mysample, prob=T, breaks=br)
```

Now let's superimpose the probability density function of the standard normal distribution, which is defined as 
$$
f(x) = \frac{1}{\sqrt{2\pi}}e^{-\frac{x^2}{2}}
$$

```r
x<-seq(-3.5, 3.5, 0.0001)
fx<-(1/sqrt(2*pi))*exp(-0.5*x^2)
hist(mysample, prob=T, breaks=br)
lines(x, fx)
```
