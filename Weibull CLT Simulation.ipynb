Recall that the Central Limit Theorem (CLT) states:

Let $X_1, X_2, \dots, X_n$ be random variables from some distribution with mean $\mu$ and variance $\sigma^2$. Then  $\bar{X}=\frac{1}{n} \sum_{i=1}^n X_i$ will be approximately normally distributed and $\frac{\bar{X}-\mu}{\sigma / \sqrt{n}} \sim N(0, 1)$.

We will use the **Weibull distribution** $X \sim \text{Weibull}(k=1.5,\lambda=1)$ as an example.

The code below generates **5,000 random observations** from this distribution.


```r
set.seed(12345)
data = rweibull(5000, 1.5, scale=1)
# Plotting a histogram of the values
breaks = seq(0, 4, 0.25)
hist(data, breaks=breaks)
```

Now let's ceate a function that computes n sample means of this Weibull distribution (data = rweibull(5000, 1.5, scale=1)) and plot them as a histogram. That is, we will generate n random datapoints from the population and compute the sample mean of each one.
The distrubution appears to be approximately normal.

```r
sample.mean <- function(data, n){
    sample <- sample(data, n) # this fucntion creates a vector of samples
    mean <- mean(sample) # computing the average/sample mean of each sample
    mean
}

# now we use the fucntion the generate 1000 sample means with n=2000 
# replicate() runs the fucntion 100 times
means <- replicate(1000, sample.mean(data, 2000)) 
hist(means, xlim=c(0.85, 0.97))
```
