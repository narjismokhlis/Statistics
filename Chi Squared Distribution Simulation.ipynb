If $X \sim N(0,1)$, then squaring the random variable gives:

$$
X^2 \sim \chi^2_1
$$

That is, squaring a standard normal random variable results in a random variable with a **chi-squared distribution with one degree of freedom**.

In this notebook, we will demonstrate this relationship by creating a random sample from the standard normal distribution, squaring the sample, and plotting the datapoints. We will then superimpose the probability density function of the $\chi^2_1$ distribution to see how well it fits our simulated data.

```r
# generating 100000 datapoints from the standard normal distribution
set.seed(12345)
mysample<-rnorm(100000) 
# squaring them to get a chi-squared distribution with one degree of freedom
newsample<-mysample^2
# checking the max and min to determine the breaks in the histogram
max(newsample)
min(newsample)
# computing the mean because n is the mean of the chi-squared distribution
mean(newsample)
```

The code below plots the simulated chi-squared datapoints, and then superimposes the probability density function.

```r
# using the values we just computed above
n=1.00062849144235
x<-seq(0, 4, 0.0001)
fx<-(1/gamma(n/2))*(1/2)^(n/2)*x^((n/2)-1)*exp((-1/2)*x)
br<-seq(7.5e-13, 32, 0.1)
hist(newsample, prob=T, breaks=br, xlim=c(0,4))
lines(x,fx)
```
